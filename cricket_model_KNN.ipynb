```python
from sklearn.neighbors import KNeighborsClassifier
# Features: [Line, Length]
X = [
    [2, 2], [3, 2], [1, 2], [4, 2],      # Good Length
    [2, 1], [3, 1], [4, 1], [-1, 1],     # Short Ball
    [2, 3], [3, 3], [1, 3], [5, 3]       # Overpitched
]

# What shot you should play
y = [
    "Forward Defense", "Cover Drive", "On Drive", "Cover Drive",     # Good Length
    "Pull Shot", "Pull Shot", "Hook Shot", "Pull Shot",              # Short Ball
    "Drive", "Cover Drive", "Straight Drive", "Cover Drive"   # Overpitched
]
model = KNeighborsClassifier(n_neighbors = 2)

model.fit(X,y)

line = int(input("Give the line, 3+ is offside, 2 is middle, 1 and - is leg side:"))
length = int(input("Give the length, short = 1, GL = 2, overpitch = 3;"))
shot = [[line, length]]

prediction = model.predict(shot)

print("Prediction:", prediction)
```
